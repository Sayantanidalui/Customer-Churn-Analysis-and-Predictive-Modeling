# Build a Logistic Regression Model for Churn Prediction

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix
from sklearn.linear_model import LogisticRegression
```

```python
data1=customer1.copy(deep=True)
```

```python
data1.info()
```

```python
data1['Churn']=data1['Churn'].map({'No':0,'Yes':1})
print(data1['Churn'])
```

```python
new_data = data1.drop('customerID',axis = 1)
```

```python
new_data=pd.get_dummies(new_data, drop_first=True)
```

```python
columns_list2=list(new_data.columns)
print(columns_list2)
```

```python
features2=list(set(columns_list2)-set(['Churn']))
print(features2)
```

```python
y=new_data['Churn'].values
print(y)
```

```python
x = new_data[features2].values
print(x)
```

```python
train_x,test_x,train_y,test_y = train_test_split(x,y,test_size=0.3, random_state=0)
```

```python
logistic = LogisticRegression(max_iter=2000,class_weight='balanced')
```

```python
logistic.fit(train_x,train_y)
```

```python
prediction = logistic.predict(test_x)
print(prediction)
```

```python
accuracy_score=accuracy_score(test_y, prediction)
print(accuracy_score)
```

```python
print('Misclassified samples: %d' % (test_y != prediction).sum())
```

```python
confusion_matrix = confusion_matrix(test_y, prediction)
print(confusion_matrix)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(test_y,prediction))
```
